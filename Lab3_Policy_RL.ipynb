# **Deterministic Policy**

```python
policy = {
    "Battery Low": "Charge",
    "Obstacle": "Turn Left",
    "Destination Reached": "Stop"
}

state = input("Enter Current State: ")

if state in policy:
    print("Action:", policy[state])
else:
    print("Unknown State")
```

# **Policy Table**

```python
states = ["S1", "S2", "S3", "S4"]

policy = {
    "S1": "Right",
    "S2": "Up",
    "S3": "Left",
    "S4": "Down"
}

for state in states:
    print(state, "→", policy[state])
```

```python
state = input("Enter Current State: ")
print(f"Action:{policy[state]}")
```

# **Stochastic Policy**

```python
import random

r = random.random()
state=["S1","S2","S3"]

print("Random Number =", round(r,3))

if r < 0.2:
    action = "Left"

elif r < 0.5:
    action = "Right"

else:
    action = "Straight"

print("Selected Action =", action)
```
