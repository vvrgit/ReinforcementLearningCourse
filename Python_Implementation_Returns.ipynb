# **Undiscounted Return**

```python
returns=[10,20,50,5,12]
total_return=0
for return_value in returns:
  total_return=total_return+return_value
print(f"total_return:{total_return}")
```

# **Discounted Return**

```python
returns=[10,20,50,5,12]
gama=0.9
total_return=0
for episode in range(len(returns)):
  return_value=returns[episode]
  total_return=total_return+(gama**episode)*return_value
print(f"total_return:{total_return}")
```

# **User Input**

```python
episodes=input("enter the number of episodes:")
episodes=int(episodes)
returns=[]
for episode in range(episodes):
  return_value=input(f"enter the return value for episode {episode+1}:")
  return_value=int(return_value)
  returns.append(return_value)
print(f"returns:{returns}")
gama=input("enter the gama value:")
gama=float(gama)

print(f"undiscounted return:{sum(returns)}")
total_return=0
for episode in range(len(returns)):
  return_value=returns[episode]
  total_return=total_return+(gama**episode)*return_value
print(f"discounted return:{total_return}")
```
