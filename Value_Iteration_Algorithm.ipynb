# **States**

```python
def env_states():
  state=["S0","S1"]
  return state
```

```python
states=env_states()
states
```

# **Actions**

```python
def env_actions():
  action=["up","down","right","left"]
  return action
```

```python
actions=env_actions()
actions
```

# **Rewards**

```python
def env_rewards():
  rewards={
      "move":-1,
      "wall":-1,
      "goal":10
  }
  return rewards
```

```python
rewards=env_rewards()
rewards
```

# **Transition**

```python
def env_transition():
  transition={
      "S0":{"up":"S0","down":"S1","right":"S0","left":"S0"},
      "S1":{"up":"S0","down":"S1","right":"goal","left":"S1"}
  }
  return transition
```

```python
transition=env_transition()
transition["S1"]["right"]
```

# **Discount Factor**

```python
def env_df():
  gama=0.9
  return gama
```

```python
gamma=env_df()
gamma
```

# **Initialization**

```python
def env_statevalues():
  V={
      "S0":0,
      "S1":0,
      "goal":0
  }
  return V
```

```python
V=env_statevalues()
V["S0"]
```

```python
theta=0
max_iter=1000
```

```python
def env_action_value():
  Q={
      "S0":{"up":0,"down":0,"right":0,"left":0},
      "S1":{"up":0,"down":0,"right":0,"left":0}
  }
  return Q
```

```python
Q=env_action_value()
Q["S1"]["right"]
```

# **Next State Estimation**

```python
def env_nextstate(state,action):
  next_state=transition[state][action]
  return next_state
```

```python
next_state=env_nextstate("S1","right")
print(f"next state={next_state}")
```

```python
def env_policy():
  p={
      "S0":"0",
      "S1":"0"
  }
  return p
```

```python
P=env_policy()
P["S0"]
```

# **Reward Estimation**

```python
def env_reward(next_state):
  if(next_state=="goal"):
    reward=rewards["goal"]
  elif(next_state=="wall"):
    reward=rewards["wall"]
  else:
    reward=rewards["move"]
  return reward
```

```python
next_state=env_nextstate("S1","right")
print(f"next state={next_state}")
reward=env_reward(next_state)
print(f"reward={reward}")
```

# **Value Iteration Algorithm**

```python
for iter in range(max_iter):
  old_V = V.copy()
  for state in states:
    for action in actions:
      next_state=env_nextstate(state,action)
      reward=env_reward(next_state)
      Q[state][action]=reward+gamma*V[next_state]
    V[state]=max(Q[state].values())
    P[state]=max(Q[state], key=Q[state].get)

  # Compute the maximum absolute difference between state values
  max_diff = max(abs(old_V[s] - V[s]) for s in states)
  if max_diff < theta:
    break
  else:
    old_V=V
print(f"Final State Value={V}")
print(f"Final Policy={P}")
```
